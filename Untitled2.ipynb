## Section 2 — Dataset Loading and Preparation

```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("/content")

columns = (
    ["engine_id", "cycle"]
    + [f"setting_{i}" for i in range(1, 4)]
    + [f"sensor_{i}" for i in range(1, 22)]
)

train_path = DATA_DIR / "train_FD001.txt"
test_path = DATA_DIR / "test_FD001.txt"

if not train_path.exists():
    raise FileNotFoundError(
        "Upload train_FD001.txt to the Colab /content folder first."
    )

train_df = pd.read_csv(
    train_path,
    sep=r"\s+",
    header=None,
    names=columns
)

print("Training dataset shape:", train_df.shape)
print("Number of engines:", train_df["engine_id"].nunique())

sensor_columns = [
    f"sensor_{i}" for i in range(1, 22)
]

variable_sensors = [
    col for col in sensor_columns
    if train_df[col].nunique() > 1
]

feature_columns = variable_sensors

print("Variable sensor features:", len(feature_columns))
print("Selected features:", feature_columns)

print("\nFirst five rows:")
display(train_df.head())

print("\nMissing values in selected features:")
print(train_df[feature_columns].isnull().sum().sum())
```

##Section 3 — Train and Evaluate Anomaly Detection Model

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report

# Prepare sensor features
X_train = train_df[feature_columns].copy()

# Train Isolation Forest on training sensor data
model = IsolationForest(
    n_estimators=200,
    contamination=0.05,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train)

# Predict anomalies in the training data
train_predictions = model.predict(X_train)

# Isolation Forest returns:
#  1  = normal observation
# -1  = anomaly
train_df["anomaly_prediction"] = train_predictions
train_df["is_anomaly"] = train_predictions == -1

anomaly_count = int(train_df["is_anomaly"].sum())
total_count = len(train_df)
anomaly_rate = anomaly_count / total_count * 100

print("Model training completed.")
print("Training observations:", total_count)
print("Selected sensor features:", len(feature_columns))
print("Detected anomalies:", anomaly_count)
print(f"Anomaly rate: {anomaly_rate:.2f}%")

print("\nAnomaly counts:")
print(train_df["is_anomaly"].value_counts().rename({
    True: "Anomalies",
    False: "Normal observations"
}))
```

## Section 4 — Evaluate on Unseen Test Data

```python
# Section 4 — Evaluate on Unseen Test Data

from pathlib import Path
import pandas as pd

test_path = Path("/content/test_FD001.txt")

if not test_path.exists():
    raise FileNotFoundError("test_FD001.txt was not found in /content.")

test_df = pd.read_csv(
    test_path,
    sep=r"\s+",
    header=None,
    names=columns
)

X_test = test_df[feature_columns].copy()

# Apply the model trained in Section 3
test_predictions = model.predict(X_test)

test_df["anomaly_prediction"] = test_predictions
test_df["is_anomaly"] = test_predictions == -1

test_anomaly_count = int(test_df["is_anomaly"].sum())
test_total_count = len(test_df)
test_anomaly_rate = test_anomaly_count / test_total_count * 100

print("Test evaluation completed.")
print("Test observations:", test_total_count)
print("Engines in test data:", test_df["engine_id"].nunique())
print("Sensor features:", len(feature_columns))
print("Detected anomalies:", test_anomaly_count)
print(f"Test anomaly rate: {test_anomaly_rate:.2f}%")

print("\nAnomaly counts:")
print(
    test_df["is_anomaly"].value_counts().rename({
        True: "Anomalies",
        False: "Normal observations"
    })
)
```

## Section 5 — Save the Trained Model and Feature Metadata

```python
# Section 5 — Save the Trained Model and Feature Metadata

from pathlib import Path
import json
import joblib

PROJECT_DIR = Path("/content/sensor-anomaly-api")
PROJECT_DIR.mkdir(parents=True, exist_ok=True)

# Save the trained model
model_path = PROJECT_DIR / "isolation_forest_model.joblib"
joblib.dump(model, model_path)

# Save feature names and model configuration
metadata = {
    "project_name": "Sensor Anomaly Detection API",
    "model_type": "IsolationForest",
    "feature_columns": feature_columns,
    "n_features": len(feature_columns),
    "n_estimators": 200,
    "contamination": 0.05,
    "random_state": 42
}

metadata_path = PROJECT_DIR / "metadata.json"
metadata_path.write_text(json.dumps(metadata, indent=2))

print("Model and metadata saved successfully.")
print("Model path:", model_path)
print("Metadata path:", metadata_path)
print("Number of features:", len(feature_columns))

# Verify that the saved model can be loaded
loaded_model = joblib.load(model_path)

print("Saved model loaded successfully:", type(loaded_model).__name__)
print("Saved model feature count:", loaded_model.n_features_in_)
```

## Section 6 — Build the FastAPI Application

```python

# Section 6 — Build the FastAPI Application

import json
from pathlib import Path

from fastapi import FastAPI
from pydantic import BaseModel, ConfigDict
import joblib
import pandas as pd

PROJECT_DIR = Path("/content/sensor-anomaly-api")

# Load the saved model and feature metadata
api_model = joblib.load(
    PROJECT_DIR / "isolation_forest_model.joblib"
)

metadata = json.loads(
    (PROJECT_DIR / "metadata.json").read_text()
)

FEATURES = metadata["feature_columns"]

app = FastAPI(
    title="Sensor Anomaly Detection API",
    description=(
        "Detect unusual sensor observations using "
        "an Isolation Forest model trained on NASA C-MAPSS FD001."
    ),
    version="1.0.0"
)

class SensorInput(BaseModel):
    model_config = ConfigDict(extra="forbid")

    # Values are defined dynamically from the saved feature list
    # in the endpoint below, so the API stays aligned with the model.
    values: list[float]

@app.get("/health")
def health():
    return {
        "status": "ok",
        "model_loaded": True,
        "n_features": len(FEATURES)
    }

@app.post("/predict")
def predict(data: SensorInput):
    if len(data.values) != len(FEATURES):
        from fastapi import HTTPException
        raise HTTPException(
            status_code=422,
            detail=f"Expected {len(FEATURES)} feature values."
        )

    row = pd.DataFrame([data.values], columns=FEATURES)
    prediction = int(api_model.predict(row)[0])
    score = float(api_model.decision_function(row)[0])

    return {
        "prediction": "anomaly" if prediction == -1 else "normal",
        "prediction_code": prediction,
        "anomaly_score": score,
        "features_used": len(FEATURES)
    }

print("FastAPI application created.")
print("Features expected:", len(FEATURES))
print("Endpoints: GET /health and POST /predict")
```

## Section 7 — Test the FastAPI Application

```python
# Section 7 — Test the FastAPI Application

from fastapi.testclient import TestClient
import numpy as np

client = TestClient(app)

# Test 1: Health endpoint
response = client.get("/health")
assert response.status_code == 200, response.text
assert response.json()["status"] == "ok"
print("PASS: Health endpoint")

# Create a sample using real sensor values
sample_values = (
    train_df[FEATURES]
    .iloc[0]
    .astype(float)
    .tolist()
)

# Test 2: Valid prediction request
response = client.post(
    "/predict",
    json={"values": sample_values}
)
assert response.status_code == 200, response.text

result = response.json()
assert result["prediction"] in ["normal", "anomaly"]
assert np.isfinite(result["anomaly_score"])

print("PASS: Valid prediction")
print("Sample response:", result)

# Test 3: Incorrect number of feature values
response = client.post(
    "/predict",
    json={"values": sample_values[:-1]}
)
assert response.status_code == 422, response.text
print("PASS: Incorrect feature count rejected")

# Test 4: Unexpected JSON field
response = client.post(
    "/predict",
    json={
        "values": sample_values,
        "unexpected": 123
    }
)
assert response.status_code == 422, response.text
print("PASS: Unexpected JSON field rejected")

print("\nAll 4 API tests passed.")
```

```python

from pathlib import Path
import json
import zipfile

PROJECT_DIR = Path("/content/sensor-anomaly-api")
PROJECT_DIR.mkdir(parents=True, exist_ok=True)

# Save the FastAPI application to app.py
app_code = '''
import json
from pathlib import Path
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, ConfigDict
import joblib
import pandas as pd

PROJECT_DIR = Path(__file__).parent
api_model = joblib.load(PROJECT_DIR / "isolation_forest_model.joblib")
metadata = json.loads((PROJECT_DIR / "metadata.json").read_text())
FEATURES = metadata["feature_columns"]

app = FastAPI(
    title="Sensor Anomaly Detection API",
    description="Detect unusual sensor observations using an Isolation Forest model trained on NASA C-MAPSS FD001.",
    version="1.0.0"
)

class SensorInput(BaseModel):
    model_config = ConfigDict(extra="forbid")
    values: list[float]

@app.get("/health")
def health():
    return {
        "status": "ok",
        "model_loaded": True,
        "n_features": len(FEATURES)
    }

@app.post("/predict")
def predict(data: SensorInput):
    if len(data.values) != len(FEATURES):
        raise HTTPException(
            status_code=422,
            detail=f"Expected {len(FEATURES)} feature values."
        )

    row = pd.DataFrame([data.values], columns=FEATURES)
    prediction = int(api_model.predict(row)[0])
    score = float(api_model.decision_function(row)[0])

    return {
        "prediction": "anomaly" if prediction == -1 else "normal",
        "prediction_code": prediction,
        "anomaly_score": score,
        "features_used": len(FEATURES)
    }
'''

(PROJECT_DIR / "app.py").write_text(app_code.strip() + "\n")

# Save project results
summary = {
    "training_observations": int(len(train_df)),
    "test_observations": int(len(test_df)),
    "number_of_features": int(len(feature_columns)),
    "training_anomalies": int(train_df["is_anomaly"].sum()),
    "training_anomaly_rate_percent": round(
        float(train_df["is_anomaly"].mean() * 100), 2
    ),
    "test_anomalies": int(test_df["is_anomaly"].sum()),
    "test_anomaly_rate_percent": round(
        float(test_df["is_anomaly"].mean() * 100), 2
    )
}

(PROJECT_DIR / "results.json").write_text(
    json.dumps(summary, indent=2)
)

# Create project README
readme = f"""# Sensor Anomaly Detection API

A machine learning API that uses Isolation Forest to identify unusual
sensor observations from the NASA C-MAPSS FD001 dataset.

## Features
- Trains an Isolation Forest anomaly detection model.
- Analyzes training and test sensor observations.
- Saves the trained model and feature metadata.
- Provides REST API endpoints using FastAPI.
- Includes automated API tests.

## Dataset
NASA C-MAPSS FD001 turbofan engine simulation data.

Training observations: {summary['training_observations']}

Test observations: {summary['test_observations']}

Selected sensor features: {summary['number_of_features']}

Training anomaly rate: {summary['training_anomaly_rate_percent']}%

Test anomaly rate: {summary['test_anomaly_rate_percent']}%

## API Endpoints

- GET /health — checks API health and model loading.
- POST /predict — predicts whether a sensor observation is normal or anomalous.

## Example Request

Send a JSON object to POST /predict with a `values` list containing exactly
{summary['number_of_features']} numeric sensor values, in the feature order
listed in metadata.json.

Example structure:

    {{
      "values": [0.0, 0.0, 0.0]
    }}

The three values above illustrate the JSON structure only. Replace them with
exactly {summary['number_of_features']} numeric values for a valid request.

## Run Locally

Install the dependencies:

    pip install fastapi uvicorn scikit-learn pandas joblib

Start the API from this project directory:

    uvicorn app:app --reload

Open the interactive API documentation at:

    http://127.0.0.1:8000/docs

## Model Details

Algorithm: Isolation Forest

Number of estimators: 200

Contamination parameter: 0.05

Random state: 42

## Project Files

- app.py
- isolation_forest_model.joblib
- metadata.json
- results.json
- README.md

## Important Note

The anomaly rates are model-generated flags, not ground-truth accuracy
measurements. The test dataset does not provide row-by-row anomaly labels.
"""

(PROJECT_DIR / "README.md").write_text(readme)

# Check that all required files exist
files_to_package = [
    "app.py",
    "isolation_forest_model.joblib",
    "metadata.json",
    "results.json",
    "README.md"
]

missing = [
    name for name in files_to_package
    if not (PROJECT_DIR / name).exists()
]

if missing:
    raise FileNotFoundError(
        f"Required project files are missing: {missing}. "
        "Run Sections 2 through 5 again if the model or metadata is missing."
    )

# Create ZIP package
zip_path = Path("/content/sensor-anomaly-api.zip")

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as archive:
    for name in files_to_package:
        archive.write(
            PROJECT_DIR / name,
            arcname=f"sensor-anomaly-api/{name}"
        )

print("Project summary:")
print(json.dumps(summary, indent=2))
print("\nAll required project files are present.")
print("ZIP package created:", zip_path)
print("Files included:", ", ".join(files_to_package))

```
